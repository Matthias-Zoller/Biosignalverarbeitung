## Aufgabe 7: Gefilterte gegen ungefilterte Daten

Butterworth-Tiefpass 4. Ordnung mit 5 Hz Grenzfrequenz. Mit `filtfilt` wird vorwärts und rückwärts gefiltert, dadurch gibt es keine Phasenverschiebung.

```python
# Setting up a butterworth filter
def butter_lowpass(cutoff, fs, order=4):
    # nyq: Nyquist Theorem
    nyq = 0.5 * fs
    b, a = sci.butter(order, cutoff / nyq, 'lowpass')
    return b, a

def butter_lowpass_filter(signal, cutoff, fs, order=4):
    #use filtfilt for a zero-phase filtering
    b, a = butter_lowpass(cutoff, fs, order)
    return sci.filtfilt(b, a, signal, axis=0)
```

```python
fs6 = (len(df6) - 1) / (df6['Time'].iloc[-1] - df6['Time'].iloc[0])

df6_filt = df6.copy()
df6_filt[['X', 'Y', 'Z']] = butter_lowpass_filter(df6[['X', 'Y', 'Z']], 5, fs6)

fig, ax = plot.subplots(figsize= [10, 4])
df6.plot(x = 'Time', y = ['X', 'Y', 'Z'], ax=ax, alpha=0.3, legend=False)
df6_filt.plot(x = 'Time', y = ['X', 'Y', 'Z'], ax=ax, color=['C0', 'C1', 'C2'])
ax.set_xlim(5, 15)
ax.set_ylabel('Acceleration in g')
ax.set_xlabel('Time in s')
ax.set_title('Aufgabe 7: ungefiltert (hell) und gefiltert, Ausschnitt 5-15 s')
plot.savefig('figures/A7_Filter.png', dpi=150)

ruhe = (df6['Time'] > 1) & (df6['Time'] < 9)
print('Std Z in Ruhe ungefiltert:', round(df6['Z'][ruhe].std() * 1000, 1), 'mg')
print('Std Z in Ruhe gefiltert:  ', round(df6_filt['Z'][ruhe].std() * 1000, 1), 'mg')
```

## Aufgabe 8: Abtastfrequenz (Mittelwert und Abweichung)

Aus den `millis()`-Zeitstempeln: $f_s = 1/\overline{\Delta t}$ und $\sigma_{f_s} \approx \sigma_{\Delta t} / \overline{\Delta t}^2$.

```python
messungen = {'A4 (12.5 Hz)': df4, 'A5 (100 Hz)': df5, 'A6 (100 Hz)': df6}

fig, axs = plot.subplots(1, 3, figsize= [11, 3.5])
for ax, (name, df) in zip(axs, messungen.items()):
    dt = df['Time'].diff().dropna()
    fs_mean = 1 / dt.mean()
    fs_std = dt.std() / dt.mean()**2
    print(f'{name}: dt = {dt.mean()*1000:.2f} ± {dt.std()*1000:.2f} ms, fs = {fs_mean:.1f} ± {fs_std:.1f} Hz')

    ax.hist(dt * 1000, bins=20)
    ax.set_title(f'{name}: {fs_mean:.1f} ± {fs_std:.1f} Hz')
    ax.set_xlabel('dt in ms')
axs[0].set_ylabel('Anzahl')
plot.tight_layout()
plot.savefig('figures/A8_Abtastfrequenz.png', dpi=150)
```

```python
#A4: sensor only delivers new values with 12.5 Hz, the rest are repeated lines
neu = df4[['X', 'Y', 'Z']].diff().abs().sum(axis=1) > 0
print('A4 neue Sensorwerte:', round(1 / df4['Time'][neu].diff().median(), 1), 'Hz')
```

## Aufgabe 9: Mobiles Messsystem (DataLogger, Batterie, Lab1Code2)

Mobiles System laut `Lab1Code2`: Sensor MMA8452Q (±8 g, ODR 200 Hz) und Qwiic OpenLog (SD-Karte), beide über I2C am Arduino.
Aufnahme von ca. 10 s, Datei `ACCdata.txt` von der SD-Karte. Sie hat keine Kopfzeile, die Werte sind schon in g (Rohwert / 256, 2 Nachkommastellen), die Zeit in ms (`millis()`).
Vergleich der Abtastfrequenz mit der stationären Messung A6 (BMA400 über USB, ODR 100 Hz).

```python
df9 = pd.read_csv(r'data/IMU_A9_mobil_Lab1Code2.txt', sep=r'\s+', names=['X', 'Y', 'Z', 'Time'])
df9['Time'] = df9['Time'].div(1000)
df9['Betrag'] = np.sqrt(df9['X']**2 + df9['Y']**2 + df9['Z']**2)

def abtastfrequenz(df):
    dt = df['Time'].diff().dropna()
    return dt, 1 / dt.mean(), dt.std() / dt.mean()**2

for name, df in {'A6 stationaer (USB)': df6, 'A9 mobil (Logger)': df9}.items():
    dt, fs_mean, fs_std = abtastfrequenz(df)
    print(f'{name}: N = {len(df)}, Dauer = {df.Time.iloc[-1] - df.Time.iloc[0]:.1f} s, '
          f'dt = {dt.mean()*1000:.2f} ± {dt.std()*1000:.2f} ms (min {dt.min()*1000:.0f}, max {dt.max()*1000:.0f}), '
          f'fs = {fs_mean:.1f} ± {fs_std:.1f} Hz')
print('A9 mittlerer Betrag:', round(df9['Betrag'].mean(), 3), 'g')
```

```python
dt6, fs6_mean, fs6_std = abtastfrequenz(df6)
dt9, fs9_mean, fs9_std = abtastfrequenz(df9)

fig, axs = plot.subplots(1, 2, figsize= [11, 3.8], gridspec_kw={'width_ratios': [1.4, 1]})
df9.plot(x = 'Time', y = ['X', 'Y', 'Z', 'Betrag'], ax=axs[0], ylabel='Acceleration in g',
         xlabel='Time in s', title='Aufgabe 9: mobiles System (Lab1Code2)')
axs[0].legend(loc='center', ncol=4, fontsize=8)
bins = np.arange(8.5, 25.5, 1)
axs[1].hist(dt6 * 1000, bins=bins, density=True, alpha=0.6,
            label=f'A6 stationaer: {fs6_mean:.1f} ± {fs6_std:.1f} Hz')
axs[1].hist(dt9 * 1000, bins=bins, density=True, alpha=0.6,
            label=f'A9 mobil: {fs9_mean:.1f} ± {fs9_std:.1f} Hz')
axs[1].set_xlabel('dt in ms')
axs[1].set_ylabel('relative Haeufigkeit')
axs[1].set_title('Abtastintervalle')
axs[1].legend(fontsize=8)
plot.tight_layout()
plot.savefig('figures/A9_mobil.png', dpi=150)
```
